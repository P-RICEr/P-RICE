```python
import glob
import pandas as pd

# Find all CSV files in the current folder (or specify a folder path like "data/*.csv")
file_paths = glob.glob("*.csv")

for path in file_paths:
  df = pd.read_csv(path)
  if "USD to PHP" in df.columns:
    df.rename(columns={"USD to PHP": "USD_to_PHP"}, inplace=True)
    df.to_csv(path, index=False)
    print(f"Updated: {path}")
```
